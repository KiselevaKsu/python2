```python
import numpy as np
import pandas as pd

#Если размерность данных >2, то используют иерархическую индексацию(мультииндекс)
# В один инекс включается несколько уровней.

index = [
    ('city_1', 2010),
    ('city_1', 2020),
    ('city_2', 2010),
    ('city_2', 2020),
    ('city_3', 2010),
    ('city_3', 2020),
]

population = [
    101,
    201,
    102,
    202,
    102,
    203,
]

pop = pd.Series(population, index = index)

print(pop)

print(pop[ [i for i in pop.index if i[1] == 2020] ]) # 2й элемент в индексе = 2020
```

```python
# MultiIndex

index = pd.MultiIndex.from_tuples(index)

pop = pop.reindex(index)
print(pop)

print(pop[:, 2020])
```

```python
pop_df = pop.unstack()
print(pop_df)

print(pop_df.stack())
```

```python
# Как можно создавать мультииндексы:
# 1. список массивов, которые задают значения индекса на каждом уровне 

i1 = pd.MultiIndex.from_arrays(
    [
        ['a','a','b','b'],
        [1,2,1,2]
    ]
)
print(i1)

# 2. список кортежей, задающих значение индекса в каждой точке

i2 = pd.MultiIndex.from_tuples(
    [
        ('a', 1),
        ('a', 2),
        ('b', 1),
        ('b', 2),
    ]
)
print(i2)

# 3. декартово произведение обычных индексов

i3 = pd.MultiIndex.from_product(
    [
        ['a','b'],
        [1,2]
    ]
)
print(i3)

# 4. описание внутреннего представления: levels, codes

i4 = pd.MultiIndex(
    levels = [
        ['a','b'],
        [1,2]
    ],
    codes = [ # как размещаются
        [0,0,1,1], # a a b b 
        [0,1,0,1] # 1 2 1 2
    ]
)
print(i4)
```

```python
# Уровням можно задавать названия

data = {
    ('city_1', 2010): 100,
    ('city_1', 2020): 200,
    ('city_2', 2010): 1001,
    ('city_2', 2020): 2001,
}

s = pd.Series(data)
print(s)

s.index.names = ['city','year']
print(s)
```

```python
index = pd.MultiIndex.from_product(
    [
        ['city_1','city_2'],
        [2010,2020]
    ],
    names = ['city','year']
)
print(index)

columns = pd.MultiIndex.from_product(
    [
        ['person_1','person_2', 'person_3'],
        ['job1','job2','job3']
    ],
    names = ['worker','job']
)

rng = np.random.default_rng(1)

data = rng.random((4,9))
print(data)

data_df = pd.DataFrame(data, index=index, columns=columns)
print(data_df)
```

```python
# Индексация и срезы(по мультииндексу)

data = {
    ('city_1', 2010): 100,
    ('city_1', 2020): 200,
    ('city_2', 2010): 1001,
    ('city_2', 2020): 2001,
    ('city_3', 2010): 10001,
    ('city_3', 2020): 20001,
}

s = pd.Series(data)
s.index.names = ['city','year']

print(s['city_1',2010])
print(s['city_1'])

print(s.loc['city_1':'city_2'])
print(s[:,2010])

print(s[s>2000])

print(s.loc[['city_1','city_3']])
```

```python
# Перегруппировка мультииндексов 

index = pd.MultiIndex.from_product(
    [
        ['a','c','b'],
        [1,2]
    ]
)

data = pd.Series(rng.random(6), index = index)
data.index.names = ['char', 'int']

print(data)
# print(data['a':'b']) - не работает тк индексы не отсортированы

data = data.sort_index()
print(data)
print(data['a':'b']) 
```

```python
index = [
    ('city_1', 2010, 1),
    ('city_1', 2010, 2),
    
    ('city_1', 2020, 1),
    ('city_1', 2020, 2),
    
    ('city_2', 2010, 1),
    ('city_2', 2010, 2),
    
    ('city_2', 2020, 1),
    ('city_2', 2020, 2),
    
    ('city_3', 2010, 1),
    ('city_3', 2010, 2),
    
    ('city_3', 2020, 1),
    ('city_3', 2020, 2),
]

population = [
    101,
    1010,
    201,
    2010,
    102,
    1020,
    202,
    2020,
    102,
    1020,
    203,
    2030,
]

pop = pd.Series(population, index=index)
print(pop)

i = pd.MultiIndex.from_tuples(index)
pop = pop.reindex(i)

print(pop)

print(pop.unstack()) # в dataframe

print(pop.unstack(level = 0)) # столбцом становится 0 индекс (city)
print(pop.unstack(level = 1))
print(pop.unstack(level = 2))
```

```python
# NumPy конкатенация

x = [1,2,3]
y = [4,5,6]
z = [7,8,9]

print(np.concatenate([x,y,z]))


x = [[1,2,3]]
y = [[4,5,6]]
z = [[7,8,9]]

print(np.concatenate([x,y,z]))
print(np.concatenate([x,y,z], axis=0))
print(np.concatenate([x,y,z], axis=1))
```

```python
# Pandas - concat

ser1 = pd.Series(['a','b','c'],index=[1,2,3])
ser2 = pd.Series(['d','e','f'],index=[4,5,6])

print(pd.concat([ser1,ser2]))

ser1 = pd.Series(['a','b','c'],index=[1,2,3])
ser2 = pd.Series(['d','e','f'],index=[1,2,6])

print(pd.concat([ser1,ser2], verify_integrity=False)) #Чтобы не повторялись
print(pd.concat([ser1,ser2], ignore_index=True)) # Свои индексы делает
print(pd.concat([ser1,ser2], keys=['x','y']))

# print(pd.concat([ser1,ser2], join='outer'))
# print(pd.concat([ser1,ser2], join='inner'))
```
