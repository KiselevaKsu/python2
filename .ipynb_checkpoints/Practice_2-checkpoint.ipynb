```python
# Написать нейросеть, которая будет складывать два небольших числа (от 0 до 10)
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

X = np.array([[a, b] for a in range(11) for b in range(11)])  # Все комбинации 0-10
y = X.sum(axis=1) 

model = Sequential()
model.add(Dense(1, input_dim=2, activation=None))

model.compile(optimizer='adam', loss='mse')

model.fit(X, y, epochs=1, verbose=1)

test = np.array([[3,5], [0,10], [7,2]])
print("Результаты:")
for inp, true_sum in zip(test, test.sum(axis=1)):
    pred = model.predict(inp.reshape(1, -1))[0][0]
    print(f"{inp[0]} + {inp[1]} = {pred:.1f} (должно быть {true_sum})")
```
