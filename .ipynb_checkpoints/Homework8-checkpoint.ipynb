```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
x1 = [1, 5, 10, 15, 20]
y1 = [1, 7, 3, 5, 11]

x2 = [1, 5, 10, 15, 20]
y2 = [4, 3, 1, 8, 12]

plt.figure()
plt.plot(x1, y1, color='red', marker='o', linestyle='-', label='Кривая 1')
plt.plot(x2, y2, color='green', marker='o', linestyle='-.', label='Кривая 2')
plt.title('График двух кривых')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(False)
plt.show()
```

```python
x1 = np.linspace(0, 10, 100)
y1 = np.sin(x1)

x2 = np.linspace(-5, 5, 100)
y2 = (x2**2)

x3 = np.linspace(-10, 10, 100)
y3 = -0.1 * (x3**2) + 10

plt.figure(figsize=(10, 8))

plt.subplot(2, 1, 1)
plt.plot(x1, y1, linestyle='-')
plt.xticks(np.arange(0, 11, 1), [f"{i:.1f}" for i in np.arange(0, 11, 1)])
plt.yticks(np.arange(-1, 2, 1))

plt.subplot(2, 2, 3)
plt.plot(x2, y2, linestyle='-')
plt.xticks(np.arange(-5, 6, 1), [f"{i:.1f}" for i in np.arange(-5, 6, 1)])
plt.yticks(np.arange(0, 26, 5))

plt.subplot(2, 2, 4)
plt.plot(x3, y3, linestyle='-')
plt.xticks(np.arange(-10, 11, 2), [f"{i:.1f}" for i in np.arange(-10, 11, 2)])
plt.yticks(np.arange(-10, 11, 2))

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 20, 1000)
y = (0.5*x - 5)**2 

fig, ax = plt.subplots()
ax.plot(x, y)
ax.axis('equal')

ax.annotate('min', xy=(10, 0), xytext=(10, 10),
            arrowprops=dict(facecolor='green'))

plt.xlabel('X')
plt.ylabel('Y')
plt.xlim(0, 20)
plt.ylim(0, 20) 
plt.grid()
plt.show()
```

```python
data = np.random.randint(0, 11, (8, 8)) 

plt.imshow(data, cmap='viridis', interpolation='nearest')

plt.colorbar()

plt.xticks(range(8))
plt.yticks(range(8))

plt.show()
```

```python
x = np.linspace(0, 5, 100) 
y = np.cos(2 * np.pi * x)

plt.plot(x, y, 'r', linewidth=2) 
plt.fill_between(x, y, 0, where=(y > 0), color='blue') 
plt.fill_between(x, y, 0, where=(y < 0), color='blue')

plt.show()
```

```python
x = np.linspace(0, 20, 100)
y = np.sin(x)

y_masked = np.ma.masked_less(y, 0) 

plt.plot(x, y_masked)
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(12, 4))

# 1) Лесенка над прямой y=x
x1 = [0, 1, 2, 3, 4, 5, 6]
y1 = [1, 2, 3, 4, 5, 6, 6]
ax[0].step(x1, y1, where='post', label='Лесенка над y=x')
ax[0].scatter(x1, [0, 1, 2, 3, 4, 5, 6], color='red')
ax[0].set_xticks(x1)
ax[0].set_yticks([0, 1, 2, 3, 4, 5, 6])
ax[0].grid(True)
ax[0].legend()

# 2) Лесенка под прямой y=x
x2 = [0, 1, 2, 3, 4, 5, 6]
y2 = [0, 1, 2, 3, 4, 5, 6]
ax[1].step(x2, y2, where='post', label='Лесенка под y=x')
ax[1].scatter(x2, [0, 1, 2, 3, 4, 5, 6], color='red')
ax[1].set_xticks(x2)
ax[1].set_yticks([0, 1, 2, 3, 4, 5, 6])
ax[1].grid(True)
ax[1].legend()

# 3) Лесенка, середина ступенки которой y=x
x3 = [0.5, 1.5, 2.5, 3.5, 4.5, 5.5, 6]
y3 = [0, 1, 2, 3, 4, 5, 6]
ax[2].step(x3, y3, where='mid', label='Лесенка с серединой на y=x')
ax[2].scatter(x3, [0, 1, 2, 3, 4, 5, 6], color='red')
ax[2].set_xticks(x3)
ax[2].set_yticks([0, 1, 2, 3, 4, 5, 6])
ax[2].grid(True)
ax[2].legend()

plt.tight_layout()
plt.show()
```
