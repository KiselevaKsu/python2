```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x1 = [1, 5, 10, 15, 20]
y1 = [1, 7, 3, 5, 11]

x2 = [1, 5, 10, 15, 20]
y2 = [4, 3, 1, 8, 12]

plt.figure()
plt.plot(x1, y1, color='red', marker='o', linestyle='-', label='Кривая 1')
plt.plot(x2, y2, color='green', marker='o', linestyle='-.', label='Кривая 2')
plt.title('График двух кривых')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(False)
plt.show()
```

```python
x1 = np.linspace(0, 10, 100)
y1 = np.sin(x1)

x2 = np.linspace(-5, 5, 100)
y2 = (x2**2)

x3 = np.linspace(-10, 10, 100)
y3 = -0.1 * (x3**2) + 10

plt.figure(figsize=(10, 8))

plt.subplot(2, 1, 1)
plt.plot(x1, y1, linestyle='-')
plt.xticks(np.arange(0, 11, 1), [f"{i:.1f}" for i in np.arange(0, 11, 1)])
plt.yticks(np.arange(-1, 2, 1))

plt.subplot(2, 2, 3)
plt.plot(x2, y2, linestyle='-')
plt.xticks(np.arange(-5, 6, 1), [f"{i:.1f}" for i in np.arange(-5, 6, 1)])
plt.yticks(np.arange(0, 26, 5))

plt.subplot(2, 2, 4)
plt.plot(x3, y3, linestyle='-')
plt.xticks(np.arange(-10, 11, 2), [f"{i:.1f}" for i in np.arange(-10, 11, 2)])
plt.yticks(np.arange(-10, 11, 2))

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(-10, 10, 100)

fig, ax = plt.subplots()
ax.plot(x, x**2)
ax.axis('equal')

ax.annotate('min', xy=(0, 0), xytext=(0, 10),
            arrowprops=dict(facecolor='green'))

plt.xlabel('X')
plt.ylabel('Y')
plt.xlim(-10, 10)
plt.ylim(0, 11) 
plt.grid()
plt.show()
```

```python
x = np.linspace(0, 5, 100) 
y = np.cos(2 * x)

plt.plot(x, y, 'r', linewidth=2) 
plt.fill_between(x, y, 0, color='blue') 

plt.show()
```

```python
x = np.linspace(0, 20, 100)
y = np.sin(x)

y_masked = np.where(y > -0.5, y, np.nan)

plt.plot(x, y_masked)
plt.show()
```

```python
x_values = np.linspace(0, 6, 7)
y_values = np.linspace(0, 6, 7)
fig, subplots = plt.subplots(1, 3, figsize=(15, 5))

step_types = ['pre', 'post', 'mid']
for subplot, step_type in zip(subplots, step_types):
    subplot.step(x_values, y_values, where=step_type, color='red', marker='o')
    subplot.set_title(f'Step Mode: {step_type}')
    subplot.grid(True)

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 12, 30)

y1 = np.exp(-0.1 * x) + 1  
y2 = np.sin(x / 2) + 2   
y3 = np.log(x + 1) + 3    

y2 += y1
y3 += y2

plt.stackplot(x, y1, y2 - y1, y3 - y2, labels=['y1', 'y2', 'y3'], colors=['red', 'blue', 'green'])
plt.legend(loc='upper left')
plt.title('Стековый график различных функций')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
tips = sns.load_dataset('tips')
day_counts = tips['day'].value_counts()
top_3 = day_counts.nlargest(3)
other = day_counts.iloc[3:].sum()
top_3['Остальное'] = other

explode = [0.1] + [0] * 3
plt.figure(figsize=(8, 8))
plt.pie(top_3, labels=top_3.index, explode=explode, autopct='%1.1f%%')
plt.title('Распределение чаевых по дням')
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(top_3, labels=top_3.index, explode=explode, wedgeprops={'width': 0.4}, autopct='%1.1f%%')
plt.title('Распределение чаевых по дням')
plt.show()
```
